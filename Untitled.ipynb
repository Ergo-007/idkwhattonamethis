```python
import sounddevice as sd

def list_audio_devices():
    print("=" * 80)
    print("                      AUDIO HARDWARE DISCOVERY                         ")
    print("=" * 80)
    
    devices = sd.query_devices()
    hostapis = sd.query_hostapis()
    
    default_input = sd.default.device[0]
    default_output = sd.default.device[1]
    
    print(f"\n[HOST APIS DETECTED]: {[api['name'] for api in hostapis]}\n")
    
    print(f"{'ID':<4} | {'Name':<36} | {'Host API':<12} | {'In':<4} | {'Out':<4} | {'Default SR'}")
    print("-" * 80)
    
    for idx, dev in enumerate(devices):
        api_name = hostapis[dev['hostapi']]['name']
        in_ch = dev['max_input_channels']
        out_ch = dev['max_output_channels']
        sr = int(dev['default_samplerate'])
        
        flags = []
        if idx == default_input:
            flags.append("DEFAULT IN")
        if idx == default_output:
            flags.append("DEFAULT OUT")
        flag_str = f" <-- ({', '.join(flags)})" if flags else ""
        
        print(f"{idx:<4} | {dev['name'][:34]:<36} | {api_name:<12} | {in_ch:<4} | {out_ch:<4} | {sr:<7} Hz{flag_str}")
        
    print("=" * 80)
    print("\nTARGET ASSIGNMENTS TO NOTE:")
    print("  1. Find your Laptop Internal Mic (usually Realtek Microphone Array via MME or WASAPI).")
    print("  2. Find your Focusrite Output (Focusrite USB Audio, Scarlett, via WASAPI or ASIO).")
    print("=" * 80)

if __name__ == "__main__":
    list_audio_devices()
```

```python
import sounddevice as sd
import numpy as np

SAMPLE_RATE = 16000  
BLOCK_SIZE = 512     

def audio_callback(indata, outdata, frames, time_info, status):
    if status:
        print(f"Status flag: {status}")
    
    
    mic1_voice = indata[:, 0] 
    mic2_noise = indata[:, 1]  

    
    rms_voice = np.sqrt(np.mean(mic1_voice**2)) + 1e-9
    rms_noise = np.sqrt(np.mean(mic2_noise**2)) + 1e-9

   
    db_voice = 20 * np.log10(rms_voice)
    db_noise = 20 * np.log10(rms_noise)


    meter = f"\r[CH1 Voice]: {db_voice:6.1f} dBFS | [CH2 Noise]: {db_noise:6.1f} dBFS"
    print(meter, end="", flush=True)

    
    outdata[:, 0] = mic1_voice
    outdata[:, 1] = mic1_voice

print("Starting audio stream. Speak into mic to test...")
with sd.Stream(channels=(2, 2), samplerate=SAMPLE_RATE, blocksize=BLOCK_SIZE, dtype='float32', callback=audio_callback):
    input("\nPress Enter to stop stream.\n")
```

```python
import sounddevice as sd
import soundfile as sf
import numpy as np

SAMPLE_RATE = 16000
BLOCK_SIZE = 1024

recorded_ch1 = []
recorded_ch2 = []

def audio_callback(indata, outdata, frames, time_info, status):
    mic1_voice = indata[:, 0]
    mic2_noise = indata[:, 1]

   
    recorded_ch1.append(mic1_voice.copy())
    recorded_ch2.append(mic2_noise.copy())

 
    outdata[:, 0] = mic1_voice
    outdata[:, 1] = mic1_voice

print("Recording live stream... Press Enter to stop.")
with sd.Stream(channels=(2, 2), samplerate=SAMPLE_RATE, blocksize=BLOCK_SIZE, dtype='float32', callback=audio_callback):
    input()
ch1_array = np.concatenate(recorded_ch1, axis=0)
ch2_array = np.concatenate(recorded_ch2, axis=0)

sf.write("mic1_voice_captured.wav", ch1_array, SAMPLE_RATE)
sf.write("mic2_noise_captured.wav", ch2_array, SAMPLE_RATE)
print("Saved mic1_voice_captured.wav and mic2_noise_captured.wav in current directory!")
```
