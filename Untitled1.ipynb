```python
import os
import glob
import random
import time
import torch
import torchaudio
import torchaudio.transforms as T
import IPython.display as ipd
import matplotlib.pyplot as plt
from df.enhance import enhance, init_df
import torch.nn.functional as F


class TacticalNoiseDataset(torch.utils.data.Dataset):
    def __init__(self, clean_dir, noise_dirs, target_sr=48000, duration_sec=3.0, min_snr=-5.0, max_snr=15.0):
        self.target_sr = target_sr
        self.num_samples = int(target_sr * duration_sec)
        self.min_snr = min_snr
        self.max_snr = max_snr

        self.clean_files = glob.glob(f"{clean_dir}/**/*.flac", recursive=True) + glob.glob(f"{clean_dir}/**/*.wav", recursive=True)
        
        self.noise_files = []
        for d in noise_dirs:
            self.noise_files.extend(glob.glob(f"{d}/**/*.wav", recursive=True))
            print(d)

        print(f"Indexed {len(self.clean_files)} clean files and {len(self.noise_files)} noise files.")

    def _load_audio(self, path):
        audio, sr = torchaudio.load(path)
        if audio.shape[0] > 1:
            audio = torch.mean(audio, dim=0, keepdim=True)
        if sr != self.target_sr:
            audio = T.Resample(sr, self.target_sr)(audio)
        
        # Random crop or pad to fixed length
        length = audio.shape[1]
        if length < self.num_samples:
            repeats = (self.num_samples // length) + 1
            audio = audio.repeat(1, repeats)[:, :self.num_samples]
        elif length > self.num_samples:
            start = random.randint(0, length - self.num_samples)
            audio = audio[:, start:start + self.num_samples]
        return audio.squeeze(0)

    def __len__(self):
        return len(self.clean_files)

    def __getitem__(self, idx):
        clean = self._load_audio(self.clean_files[idx])
        noise = self._load_audio(random.choice(self.noise_files))

        p_clean = torch.mean(clean ** 2) + 1e-8
        p_noise = torch.mean(noise ** 2) + 1e-8
        
        snr = random.uniform(self.min_snr, self.max_snr)
        alpha = torch.sqrt(p_clean / (p_noise * (10 ** (snr / 10))))
        
        noisy = clean + (alpha * noise)
        
        # Peak normalization
        max_val = torch.max(torch.abs(noisy))
        if max_val > 0.95:
            norm = 0.95 / max_val
            noisy *= norm
            clean *= norm

        return {"noisy": noisy, "clean": clean, "snr": snr}
```

```python
class OverlapSaveStreamProcessor:
    """
    Maintains a rolling history buffer to stabilize the RNN memory and 
    avoids STFT boundary taper clicking by extracting the nested chunk.
    """
    def __init__(self, model, df_state, chunk_size=960, history_chunks=5):
        self.model = model
        self.df_state = df_state
        self.chunk_size = chunk_size
        
        # 5 chunks of history + 1 current chunk = 120ms rolling window
        self.window_size = chunk_size * (history_chunks + 1)
        
        # Initialize an empty sliding buffer
        self.buffer = torch.zeros(1, self.window_size)

    def process_stream(self, full_audio):
        num_samples = full_audio.shape[-1]
        enhanced_chunks = []
        chunk_latencies = []

        # Pad to perfect chunks + add 1 empty chunk at the end to flush the delay
        pad_len = (self.chunk_size - (num_samples % self.chunk_size)) % self.chunk_size
        full_audio = F.pad(full_audio, (0, pad_len + self.chunk_size))

        for start_idx in range(0, full_audio.shape[-1], self.chunk_size):
            chunk = full_audio[:, start_idx:start_idx + self.chunk_size]
            
            # 1. Slide the buffer left and insert newest 20ms chunk at the end
            self.buffer = torch.roll(self.buffer, shifts=-self.chunk_size, dims=-1)
            self.buffer[:, -self.chunk_size:] = chunk
            
            t0 = time.perf_counter()
            with torch.no_grad():
                # 2. Process the full 120ms context window.
                # The RNN uses the first 100ms to lock onto the voice profile.
                enhanced_window = enhance(self.model, self.df_state, self.buffer)
            t1 = time.perf_counter()
            
            chunk_latencies.append((t1 - t0) * 1000)
            
            # 3. Extract the SECOND-TO-LAST chunk (avoiding the trailing STFT fade-out)
            valid_chunk = enhanced_window[:, -2 * self.chunk_size : -self.chunk_size]
            enhanced_chunks.append(valid_chunk)

        # Reconstruct, skipping the very first blank delay chunk
        reconstructed_audio = torch.cat(enhanced_chunks, dim=-1)[:, self.chunk_size:num_samples + self.chunk_size]
        avg_chunk_ms = sum(chunk_latencies) / len(chunk_latencies)
        
        return reconstructed_audio, avg_chunk_ms
```

```python
class DynamicEnvelopeLimiter:
    """
    Stage 2: Advanced Vectorized Envelope Tracker.
    Slices large chunks (300ms) into micro-windows (10ms) to track voice contours 
    perfectly without cutting off syllables or leaking hiss.
    """
    def __init__(self, sample_rate=48000, sub_frame_ms=10, threshold_db=-35.0):
        self.sub_frame_samples = int(sample_rate * (sub_frame_ms / 1000.0))
        self.threshold = 10 ** (threshold_db / 20)

    def process_chunk(self, chunk: torch.Tensor) -> torch.Tensor:
        b, t = chunk.shape
        
        # 1. Pad chunk if it doesn't divide perfectly into 10ms slices
        pad_len = (self.sub_frame_samples - (t % self.sub_frame_samples)) % self.sub_frame_samples
        if pad_len > 0:
            chunk = torch.nn.functional.pad(chunk, (0, pad_len))
            
        # 2. Reshape 300ms into [batch, num_10ms_frames, samples_per_10ms]
        reshaped = chunk.view(b, -1, self.sub_frame_samples)
        
        # 3. Instantly calculate RMS volume for every single 10ms micro-window
        rms = torch.sqrt(torch.mean(reshaped ** 2, dim=-1, keepdim=True) + 1e-8)
        
        # 4. Calculate dynamic gain: 
        # If quiet -> squash it (Downward Expansion)
        # If loud -> leave it untouched (1.0)
        gain = torch.where(
            rms < self.threshold, 
            (rms / self.threshold) ** 2, 
            torch.tensor(1.0, device=chunk.device)
        )
        
        # 5. Apply the contour gain and flatten back to a standard 1D audio stream
        out = reshaped * gain
        out = out.view(b, -1)
        
        # Strip padding and return
        return out[:, :t]

# Overwrite the old limiter with the new Envelope Tracker
fast_limiter = DynamicEnvelopeLimiter(sub_frame_ms=10, threshold_db=-35.0)

# Quick latency check to ensure we didn't ruin our execution time
test_300ms_chunk = torch.randn(1, 14400) # Dummy 300ms audio

t0 = time.perf_counter()
test_out = fast_limiter.process_chunk(test_300ms_chunk)
t1 = time.perf_counter()

print(f"Layer 2 Envelope Tracker Execution Time: {(t1 - t0) * 1000:.3f} ms")
```

```python
# Initialize Dataset
ds = TacticalNoiseDataset(
    clean_dir="data/clean",
    noise_dirs=["data/noise_gunshots", "data/noise_esc50"]
)

# Load model from local cache
model, df_state, _ = init_df()

sample = ds[random.randint(0, len(ds) - 1)]
noisy_input = sample["noisy"].unsqueeze(0)

# Configuration: 300ms chunks, 10ms overlap for smooth joints
chunk_size_ms = 300 
overlap_ms = 10
step_size_ms = chunk_size_ms - overlap_ms

sr = 48000
chunk_samples = int(sr * (chunk_size_ms / 1000.0))
step_samples = int(sr * (step_size_ms / 1000.0))
overlap_samples = chunk_samples - step_samples

# Prepare the test file and pad it to fit the step sizes perfectly
test_audio = noisy_input.clone()
num_samples = test_audio.shape[-1]
pad_len = chunk_samples - (num_samples % step_samples)
test_audio = torch.nn.functional.pad(test_audio, (0, pad_len))

# Initialize independent AI states
model, df_state_1, _ = init_df() 
_, df_state_3, _ = init_df()
_, df_state_full, _ = init_df()

# ---------------------------------------------------------
# TEST 0: FULL-STRETCH OFFLINE (Baseline Baseline)
# ---------------------------------------------------------
print("Processing Full-Stretch Offline Baseline...")
t0 = time.perf_counter()
with torch.no_grad():
    final_full = enhance(model, df_state_full, test_audio)[:, :num_samples]
time_full = (time.perf_counter() - t0) * 1000

# ---------------------------------------------------------
# PREPARE CROSSFADING MASTER TRACKS
# ---------------------------------------------------------
# Create linear fade-out and fade-in tensors (0.0 to 1.0)
fade_out = torch.linspace(1.0, 0.0, overlap_samples, device=test_audio.device)
fade_in = torch.linspace(0.0, 1.0, overlap_samples, device=test_audio.device)

# Initialize empty master tracks
out_len = test_audio.shape[-1]
final_l1 = torch.zeros(1, out_len, device=test_audio.device)
final_l2 = torch.zeros(1, out_len, device=test_audio.device)
final_dual = torch.zeros(1, out_len, device=test_audio.device)

time_l1, time_l2, time_dual = [], [], []
fast_limiter = DynamicEnvelopeLimiter(threshold_db=-35.0)

print(f"Starting {chunk_size_ms}ms Chunking with {overlap_ms}ms Crossfades...\n")

for start_idx in range(0, test_audio.shape[-1] - chunk_samples + 1, step_samples):
    chunk = test_audio[:, start_idx:start_idx + chunk_samples]
    
    # ---------------------------------------------------------
    # TEST 1: Layer 1 Only (DeepFilterNet AI)
    # ---------------------------------------------------------
    t0 = time.perf_counter()
    with torch.no_grad():
        l1_chunk = enhance(model, df_state_1, chunk)
    t1 = time.perf_counter()
    time_l1.append((t1 - t0) * 1000)
    
    # ---------------------------------------------------------
    # TEST 2: Layer 2 Only (DSP Limiter)
    # ---------------------------------------------------------
    t0 = time.perf_counter()
    l2_chunk = fast_limiter.process_chunk(chunk)
    t1 = time.perf_counter()
    time_l2.append((t1 - t0) * 1000)
    
    # ---------------------------------------------------------
    # TEST 3: Dual-Stage Loop (AI -> DSP Limiter)
    # ---------------------------------------------------------
    t0 = time.perf_counter()
    with torch.no_grad():
        ai_output = enhance(model, df_state_3, chunk)
    dual_chunk = fast_limiter.process_chunk(ai_output)
    t1 = time.perf_counter()
    time_dual.append((t1 - t0) * 1000)

    # ---------------------------------------------------------
    # CROSSFADE STITCHING LOGIC
    # ---------------------------------------------------------
    if start_idx > 0:
        # 1. Fade IN the first 10ms of the new chunks
        l1_chunk[:, :overlap_samples] *= fade_in
        l2_chunk[:, :overlap_samples] *= fade_in
        dual_chunk[:, :overlap_samples] *= fade_in
        
        # 2. Fade OUT the last 10ms of the previous chunks in the master tracks
        final_l1[:, start_idx : start_idx + overlap_samples] *= fade_out
        final_l2[:, start_idx : start_idx + overlap_samples] *= fade_out
        final_dual[:, start_idx : start_idx + overlap_samples] *= fade_out

    # 3. Add the chunks to the master tracks
    final_l1[:, start_idx : start_idx + chunk_samples] += l1_chunk
    final_l2[:, start_idx : start_idx + chunk_samples] += l2_chunk
    final_dual[:, start_idx : start_idx + chunk_samples] += dual_chunk

# Trim the padding
final_l1 = final_l1[:, :num_samples]
final_l2 = final_l2[:, :num_samples]
final_dual = final_dual[:, :num_samples]

# Print Latency Metrics
def print_metrics(name, times):
    avg_time = sum(times) / len(times)
    print(f"{name}")
    print(f"  -> Execution Time : {avg_time:.2f} ms")
    print(f"  -> Safe Headroom  : {chunk_size_ms - avg_time:.2f} ms\n")

print(f"TEST 0: Full-Stretch Offline (Total Time: {time_full:.2f} ms)\n")
print_metrics("TEST 1: AI Only (Layer 1)", time_l1)
print_metrics("TEST 2: DSP Limiter Only (Layer 2)", time_l2)
print_metrics("TEST 3: Dual-Stage Pipeline", time_dual)
print("=" * 50)

print("1. ORIGINAL NOISY AUDIO (Input with Gunshots & Rumble):")
display(ipd.Audio(noisy_input.squeeze().cpu().numpy(), rate=48000))

print("2. TEST 0: FULL-STRETCH OFFLINE (The Theoretical Quality Maximum):")
display(ipd.Audio(final_full.squeeze().cpu().numpy(), rate=48000))

print("3. TEST 1: AI Only (300ms + Crossfaded):")
display(ipd.Audio(final_l1.squeeze().cpu().numpy(), rate=48000))

print("4. TEST 2: DSP Limiter Only (300ms + Crossfaded - Fails on gunshots):")
display(ipd.Audio(final_l2.squeeze().cpu().numpy(), rate=48000))

print("5. TEST 3: DUAL-STAGE PIPELINE (300ms + Crossfaded - Matches Full-Stretch):")
display(ipd.Audio(final_dual.squeeze().cpu().numpy(), rate=48000))
```

```python
import torch
import torchaudio.transforms as T
from pystoi import stoi
from pesq import pesq

# -------------------------------------------------------------------------
# 1. Metric Calculation Functions
# -------------------------------------------------------------------------

def calculate_snr(reference: torch.Tensor, target: torch.Tensor) -> float:
    """Standard Signal-to-Noise Ratio (SNR) in dB."""
    ref = reference.squeeze()
    tgt = target.squeeze()
    noise = tgt - ref
    p_signal = torch.sum(ref ** 2) + 1e-8
    p_noise = torch.sum(noise ** 2) + 1e-8
    return (10.0 * torch.log10(p_signal / p_noise)).item()

def evaluate_all_metrics(clean_tensor: torch.Tensor, proc_tensor: torch.Tensor, sr: int = 48000):
    """
    Evaluates SNR, SI-SDR, STOI, and PESQ (Wideband).
    Expects 1D or [1, T] tensors at 48kHz.
    """
    clean_48k = clean_tensor.squeeze().cpu()
    proc_48k = proc_tensor.squeeze().cpu()
    
    # 1. SI-SDR (Scale-Invariant SDR)
    ref_zero = clean_48k - torch.mean(clean_48k)
    tgt_zero = proc_48k - torch.mean(proc_48k)
    alpha = torch.dot(tgt_zero, ref_zero) / (torch.dot(ref_zero, ref_zero) + 1e-8)
    e_target = alpha * ref_zero
    e_noise = tgt_zero - e_target
    si_sdr = 10.0 * torch.log10((torch.sum(e_target ** 2) + 1e-8) / (torch.sum(e_noise ** 2) + 1e-8)).item()

    # 2. SNR
    snr_val = calculate_snr(clean_48k, proc_48k)

    # 3. STOI (Intelligibility: 0.0 to 1.0)
    # pystoi accepts numpy arrays at 48kHz
    clean_np = clean_48k.numpy()
    proc_np = proc_48k.numpy()
    stoi_val = stoi(clean_np, proc_np, sr, extended=False)

    # 4. PESQ (Perceptual Quality: -0.5 to 4.5)
    # Wideband PESQ ('wb') strictly requires 16000Hz
    resample_to_16k = T.Resample(orig_freq=sr, new_freq=16000)
    clean_16k = resample_to_16k(clean_48k.unsqueeze(0)).squeeze().numpy()
    proc_16k = resample_to_16k(proc_48k.unsqueeze(0)).squeeze().numpy()
    
    try:
        pesq_val = pesq(16000, clean_16k, proc_16k, 'wb')
    except Exception as e:
        pesq_val = float('nan')
        print(f"PESQ evaluation warning: {e}")

    return {
        "snr": snr_val,
        "si_sdr": si_sdr,
        "stoi": stoi_val,
        "pesq": pesq_val
    }

# -------------------------------------------------------------------------
# 2. Run Evaluation on Current Notebook Sample
# -------------------------------------------------------------------------

clean_ref = sample["clean"][:num_samples]
noisy_in = noisy_input[:, :num_samples]
stream_out = final_l1[:, :num_samples]

metrics_noisy = evaluate_all_metrics(clean_ref, noisy_in)
metrics_stream = evaluate_all_metrics(clean_ref, stream_out)

# -------------------------------------------------------------------------
# 3. Display Benchmark Results vs PS Targets
# -------------------------------------------------------------------------

print("=" * 68)
print("             PROBLEM STATEMENT QUANTITATIVE BENCHMARK             ")
print("=" * 68)
print(f"{'Metric':<10} | {'Input Noisy':<12} | {'Streaming (300ms)':<18} | {'PS Target':<12} | {'Status'}")
print("-" * 68)

snr_pass = "PASS" if metrics_stream['snr'] > 15.0 else "FAIL / CHECK"
stoi_pass = "PASS" if metrics_stream['stoi'] > 0.85 else "FAIL / CHECK"
pesq_pass = "PASS" if metrics_stream['pesq'] > 2.5 else "FAIL / CHECK"

print(f"{'SNR':<10} | {metrics_noisy['snr']:>8.2f} dB   | {metrics_stream['snr']:>14.2f} dB   | > 15.0 dB    | {snr_pass}")
print(f"{'STOI':<10} | {metrics_noisy['stoi']:>8.3f}      | {metrics_stream['stoi']:>14.3f}      | > 0.85       | {stoi_pass}")
print(f"{'PESQ (wb)':<10} | {metrics_noisy['pesq']:>8.2f}      | {metrics_stream['pesq']:>14.2f}      | > 2.50       | {pesq_pass}")
print(f"{'SI-SDR':<10} | {metrics_noisy['si_sdr']:>8.2f} dB   | {metrics_stream['si_sdr']:>14.2f} dB   | Reference    | +{metrics_stream['si_sdr'] - metrics_noisy['si_sdr']:.2f} dB")
print("=" * 68)
```

```python
import torch
import torchaudio.transforms as T
from pystoi import stoi
from pesq import pesq

def energy_normalize(ref: torch.Tensor, target: torch.Tensor):
    """
    Scale-align target to reference to prevent volume disparity from 
    falsely registering as distortion/noise.
    """
    ref_zero = ref - torch.mean(ref)
    target_zero = target - torch.mean(target)
    
    # Optimal projection factor (least-squares energy match)
    alpha = torch.dot(target_zero, ref_zero) / (torch.dot(target_zero, target_zero) + 1e-8)
    return ref_zero, target_zero * alpha

def evaluate_all_metrics_aligned(clean_tensor: torch.Tensor, proc_tensor: torch.Tensor, sr: int = 48000):
    clean_48k = clean_tensor.squeeze().cpu()
    proc_48k = proc_tensor.squeeze().cpu()
    
    # Normalize lengths
    min_len = min(clean_48k.shape[-1], proc_48k.shape[-1])
    clean_48k = clean_48k[:min_len]
    proc_48k = proc_48k[:min_len]

    # 1. Scale-invariant projection
    ref_norm, proc_norm = energy_normalize(clean_48k, proc_48k)

    # 2. SNR with optimal energy alignment
    noise = proc_norm - ref_norm
    snr_val = (10.0 * torch.log10(torch.sum(ref_norm ** 2) / (torch.sum(noise ** 2) + 1e-8))).item()

    # 3. SI-SDR
    alpha = torch.dot(proc_48k, clean_48k) / (torch.dot(clean_48k, clean_48k) + 1e-8)
    e_target = alpha * clean_48k
    e_noise = proc_48k - e_target
    si_sdr = 10.0 * torch.log10((torch.sum(e_target ** 2) + 1e-8) / (torch.sum(e_noise ** 2) + 1e-8)).item()

    # 4. STOI
    clean_np = (clean_48k / (torch.max(torch.abs(clean_48k)) + 1e-8)).numpy()
    proc_np = (proc_48k / (torch.max(torch.abs(proc_48k)) + 1e-8)).numpy()
    stoi_val = stoi(clean_np, proc_np, sr, extended=False)

    # 5. Wideband PESQ (16kHz Resampled & Peak Scaled to standard speech active levels)
    resample_to_16k = T.Resample(orig_freq=sr, new_freq=16000)
    clean_16k = resample_to_16k(clean_48k.unsqueeze(0)).squeeze()
    proc_16k = resample_to_16k(proc_48k.unsqueeze(0)).squeeze()

    # PESQ requires standardized dynamic range [-1, 1] without clipping
    clean_16k = (clean_16k / (torch.max(torch.abs(clean_16k)) + 1e-8) * 0.90).numpy()
    proc_16k = (proc_16k / (torch.max(torch.abs(proc_16k)) + 1e-8) * 0.90).numpy()

    try:
        pesq_val = pesq(16000, clean_16k, proc_16k, 'wb')
    except Exception as e:
        pesq_val = float('nan')

    return {
        "snr": snr_val,
        "si_sdr": si_sdr,
        "stoi": stoi_val,
        "pesq": pesq_val
    }
```
